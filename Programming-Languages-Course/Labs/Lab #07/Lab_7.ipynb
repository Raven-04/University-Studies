```python
coordinates_of_cities = [["Manama", complex(50.5822436, 26.2235041)], 
                         ["Cairo", complex(31.2357257, 30.0443879) ], 
                         ["Tehran", complex(51.3896004, 35.6892523)], 
                         ["Amman", complex(35.9239625, 31.9515694)], 
                         ["Muscat", complex(58.3829448, 23.5882019)],
                         ["Abu Dhabi", complex(54.3774014, 24.4538352)]
                        ]
```

```python
for i in range(0, len(coordinates_of_cities)):
    print(coordinates_of_cities[i][0] + ": " + str(coordinates_of_cities[i][1]))
```

```python
min_distance = math.inf
closest_cities = ("", "")
```

```python
for i in range(len(coordinates_of_cities)):
    for j in range(i + 1, len(coordinates_of_cities)):
        distance = abs(coordinates_of_cities[i][1] - coordinates_of_cities[j][1])
        
        if distance < min_distance:
            min_distance = distance
            closest_cities = (coordinates_of_cities[i][0], coordinates_of_cities[j][0])
```

```python
print("The closest cities are:", closest_cities[0], "and", closest_cities[1])
print("Distance:", min_distance)
```

```python
def dots2front(s):
    count_dots = s.count(".")
    s_without_dots = s.replace(".", "")
    return "." * count_dots + s_without_dots
```

```python
print(dots2front("this.is.a.test"))
print(dots2front("1.23.456.7.."))
```

```python
# Valid Email
email1 = "Potato_1@example.com"
print(email1.split('@'))

# Invalid Email
email2 = "MOMO.com"
print(email2.split('@'))
```

```python
def check_valid_email(email):
    email = email.split('@')
    
    if len(email) != 2:
        print("Invalid Email")
        return

    print(email)

    digit = False
    special_char = False
    domain = False
    
    if len(email[0]) < 6:
        print("Invalid Email")
        return

    for i in range(len(email[0])):
        if digit and special_char:
            break
        
        if email[0][i].isdigit():
            digit = True
        elif email[0][i] in ['-', '_', '.', '#']:
            special_char = True

    
    if (email[1][-3:] in ["com", "edu", "org"]):
        domain = True

    if digit and special_char and domain:
        print("Valid Email")
    else:
        print("Invalid Email")
```

```python
check_valid_email(email1)
check_valid_email(email2)
```
