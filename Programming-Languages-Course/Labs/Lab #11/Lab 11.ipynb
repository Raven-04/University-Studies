Part A

```python
def RandomCipher():
    alphabet = "abcdefghijklmnopqrstuvwxyz"
    shuffled = "".join(random.sample(alphabet, len(alphabet)))
    return dict(zip(alphabet, shuffled))
```

```python
cipher = RandomCipher()
print(cipher)
```

Part B

```python
def Encode(txt, cipher):
    return "".join(cipher.get(c, c) for c in txt)
```

```python
encrypted = Encode("hello, world!", cipher)
print(encrypted)
```

Part C

```python
def Decode(encrypted, cipher):
    reversed_cipher = {v: k for k, v in cipher.items()}
    return "".join(reversed_cipher.get(c, c) for c in encrypted)
```

```python
decrypted = Decode(encrypted, cipher)
print(decrypted)
```

```python
def fun(lst1, lst2, lst3):
    elements = []
    for x, y, z in zip(lst1, lst2, lst3):
        element = x + " from " + z + " lives in " + y
        elements.append(element)
    return elements
```

```python
print(fun(["Ahmed", "John", "Zina"], ["Dubai", "Sharjah", "Al Ain"], ["COE", "CMP", "ELE"]))
```

```python
def enumerate(lst, start = 0, step = 1):
    return [((i * step) + start, lst[i]) for i in range(len(lst))]
```

```python
enumerate(['A', 'B', 'C'])
```

```python
enumerate(['A', 'B', 'C'], 5)
```

```python
enumerate(['A', 'B', 'C'], 4, 2)
```

Part A

```python
def apply_all(fun, lst):
    if not lst:
        return []
    return [fun(lst[0])] + apply_all(fun, lst[1:])
```

```python
apply_all(sqrt, [4, 9, 16, 25])
```

Part B

```python
apply_all(lambda x : x.isdigit(), ["1234", "hellow", "class23"])
```

```python
def PI(precision):
    getcontext().prec = precision
    def generatePI(n):
        pi = Decimal(3)
        for k in range(1, n):
            d1 = 2 * k
            d2 = d1 + 1
            d3 = d1 + 2

            term = Decimal(4) / (Decimal(d1) * Decimal(d2) * Decimal(d3))

            if k % 2 == 0:
                pi -= term
            else:
                pi += term

        return pi
    return generatePI
```

```python
REAL_PI = Decimal('3.14159265358979323846264338327950288419716939937510') 
pi = PI(precision=20)
for n in range(5,51,5):
    print('n =', n, '\t π =', res:=pi(n), '\t Δ =', abs(res-REAL_PI))
```
