```python
def convert_number(item):
    if item[:2] == ("0b"):
        print(f"{item} is base 2")       
        return int(item, 2)
    elif item[:2] == ("0o"):
        print(f"{item} is base 8")
        return int(item, 8)
    elif item[:2] == ("0x"):
        print(f"{item} is base 16")
        return int(item, 16)
    else:
        print(f"{item} is base 10")
        return int(item)
```

```python
lst = ['0b1010', '0o10', '0xFF', '25']
for item in lst:
    num = convert_number(item)
    print(f"Conversion Result:")
    print(f"Decimal     :{num}")
    print(f"Binary      :{bin(num)}")
    print(f"Octal       :{oct(num)}")
    print(f"Hexadecimal :{hex(num)}\n")
```

```python
def sort_by_decimal_count(numbers):
    unsorted_lst = []

    for num in numbers:
        decimal_count = len(str(num).split(".")[1]) if "." in str(num) else 0
        unsorted_lst.append([decimal_count, num])
    
    sorted_lst = sorted(unsorted_lst, key=lambda x: x[0], reverse=False)

    return sorted_lst
```

```python
num_lst = [12.345, 3.14, 7.1, 5.6789, 2]
sorted_lst = sort_by_decimal_count(num_lst)

print(f"Numbers sorted by decimal digits:")

for i in sorted_lst:
    print(f"{i[1]} -> Decimal digits: {i[0]}")
```

Part A

```python
def member(potential_member, group):
    if not group:
        return False
    if potential_member == group[0]:
        return True
    return member(potential_member, group[1:])
```

```python
member(2, [1, 3, 5])
```

```python
member(4, [1, 2, 3, 4, 5])
```

Part B

```python
def find(potential_member, group, index=0):
    if not group:
        return -1
    if potential_member == group[0]:
        return index 
    return find(potential_member, group[1:], index + 1)
```

```python
find(2, [1, 3, 5])
```

```python
find(4, [1, 2, 3, 4, 5])
```

```python
def flattenLst(nested_lst):
    if not nested_lst:
        return []
    if isinstance(nested_lst[0], list):
        return flattenLst(nested_lst[0]) + flattenLst(nested_lst[1:])
    else:
        return [nested_lst[0]] + flattenLst(nested_lst[1:])
```

```python
flattenLst([1, [2, 3], [4, [5, 6]]])
```
