```python
def lowercase(text): #convert the text to lowercase
    return text.lower()
```

```python
def remove_punctuation(text): #remove punctuation from the text
    punctuation = ".,!?;:'\"()[]{}"
    return "".join(char for char in text if char not in punctuation) # could also use text.strip(punctuation) instead
```

```python
def remove_numbers(text): #remove digits from the text
    digits = "0123456789"
    return "".join(num for num in text if num not in digits) # could also use text.strip(digits) instead
```

```python
def normalize_spaces(text): #remove extra space 
    return " ".join(text.split()) # or you could use text.strip() instead
```

```python
def tokenize(text): #tokenize the text
    return text.split()
```

```python
STOPWORDS = ["is", "a", "the", "this", "of", "and", "to", "in", "it"]
```

```python
def remove_stopwords(tokens): #remove STOPWORDS
    return [token for token in tokens if token not in STOPWORDS]
```

```python
def word_frequency(tokens): # calculate word frequency and save into a dictionary
    frequency = {}
    for token in tokens:
        if token in frequency:
            frequency[token] += 1
        else:
            frequency[token] = 1
    return frequency
```

```python
POSITIVE = ["good", "great", "awesome", "love", "excellent"]
NEGATIVE = ["bad", "terrible", "hate", "awful", "poor"]
```

```python
def sentiment_score(tokens): #calulate sentiment score. 
    score = 0
    for token in tokens:
        if token in POSITIVE:
            score += 1
        elif token in NEGATIVE:
            score -= 1
    return score
```

```python
def sentiment_label(score): #return sentiment label : positive, negative, or neutral
    if score > 0:
        return "Positive"
    elif score < 0:
        return "Negative"
    else:
        return "Neutral"
```

```python
def nlp_pipeline(text):
    # Cleaning
    text = lowercase(text)
    text = remove_punctuation(text)
    text = remove_numbers(text)
    text = normalize_spaces(text)

    # Tokenization
    tokens = tokenize(text)

    # Stopwords
    tokens = remove_stopwords(tokens)

    # Analysis
    freq = word_frequency(tokens)
    score = sentiment_score(tokens)
    sentiment = sentiment_label(score)

    return {
        "tokens": tokens,
        "frequency": freq,
        "sentiment_score": score,
        "sentiment": sentiment
    }
```

```python
text = "I love Python!!! It is GREAT and awesome.  But I hate indentation ."

result = nlp_pipeline(text)

print("Tokens:", result["tokens"])
print("Frequency:", result["frequency"])
print("Sentiment Score:", result["sentiment_score"])
print("Sentiment:", result["sentiment"])
```
