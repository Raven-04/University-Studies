Part a

```python
f_lst = [98, 86, 77, 11, 91, 25, 50]
```

```python
def convertToCel(f):
    c = (f - 32) * 5/9
    return c
```

```python
lstCel1 = list(map(convertToCel, f_lst))
print(lstCel1)
```

Part b

```python
lambda_f_to_c = lambda f: (f - 32) * 5/9
print(list(map(lambda_f_to_c, f_lst)))
```

```python
list(filter(lambda c: c < 0, lstCel1))
```

Part c

```python
lstCel2 = [convertToCel(fahren) for fahren in f_lst]
print(lstCel2)
```

```python
negative_c_lst = [c for c in lstCel2 if c < 0]
print(negative_c_lst)
```

Part d

```python
from functools import reduce
avg = reduce(lambda x, y: x + y, lstCel1) / len(lstCel1)
avg
```

```python
std = (reduce(lambda x, y: x + (y - avg) ** 2, lstCel1, 0) / len(lstCel1)) ** 0.5
std
```

```python
def selectionSort(lst):
  for i, v in enumerate(lst):
    minidx = reduce(lambda x, y: (x[1]<y[1])*x + (y[1]<=x[1])*y, enumerate(lst[i:]))
    minidx = minidx[0] + i
    if i != minidx: lst[i], lst[minidx] = lst[minidx], lst[i]
  return lst
```

```python
lst = [64, 25, 12, 22, 11]
selectionSort(lst)
```
