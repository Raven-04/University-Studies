```python
import math
```

```python
a = 5
b = 6
c = 1 
d = b**2 - 4*a*c
```

```python
sol1 = (-b + math.sqrt(d)) / (2*a)
sol2 = (-b - math.sqrt(d)) / (2*a)
```

```python
print("Solution 1:", sol1)
print("Solution 2:", sol2)
```

```python
sum_odd = 0
sum_even = 0

for num in range(1, 1001):
    if num % 2 == 0:
        sum_even += num
    else:
        sum_odd += num

print(f"Sum of odd numbers: {sum_odd}")
print(f"Sum of even numbers: {sum_even}")
```

```python
sum_odd = 0
sum_even = 0

for num in range(1, 1001, 2):
    sum_odd += num

for num in range(2, 1001, 2):
    sum_even += num

print(f"Sum of odd numbers: {sum_odd}")
print(f"Sum of even numbers: {sum_even}")
```

```python
grades = [40, 86.5, 67.8, 55, 43.7, 85]

grades.extend([96, 71])

grades.sort()
print("Sorted Grades:", grades)

fail_grades = [grade for grade in grades if grade <= 50]

print("Fail Grades:")
for grade in fail_grades:
    print(grade)
```
