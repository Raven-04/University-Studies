Part a

```python
def myJoin(lst, sep):
    if len(lst) == 0:
        return ""
    
    if len(lst) == 1:
        return lst[0]
    
    return lst[0] + sep + myJoin(lst[1:], sep)
```

```python
print(myJoin(["abc", "def", "gh"], "-"))
print(myJoin(["one", "two", "three"], ","))
```

Part b

```python
def flattenJoin(lst, sep):
    if len(lst) == 0:
        return ""
    
    if len(lst) == 1:
        return flattenJoin(lst[0], sep) if isinstance(lst[0], list) else lst[0]
    
    first = flattenJoin(lst[0], sep) if isinstance(lst[0], list) else lst[0]
    rest = flattenJoin(lst[1:], sep)
    return first + sep + rest if rest else first
```

```python
print(flattenJoin([["abc", ["def"]], "gh"], "-"))
print(flattenJoin(["one", ["two", ["three"]]], ","))
```

```python
lst = [("Ahmad","COE",3.5),("Bilal","CMP",3.8), ("Rania","CMP",2.4),("Khalid","COE",2.8),
("Ali","COE",3.75),("John","CMP",3.3),("Ehsan","COE",2.5)]

print(f"COE Students: ")
for name, major, gpa in lst:
    if major == "COE":
        print(f"{name:<8} {gpa}")

count = 0
print(f"\nStudents with <3 GPA")
for name, major, gpa in lst:
    if gpa < 3:
        count += 1
        print(f"{name:<8} {gpa}")
print(f"Total = {count}")
```

```python
script = "The best programs are written so that computing machines can perform them quickly and so that human beings can understand them clearly. A programmer is ideally an essayist who works with traditional aesthetic and literary forms as well as mathematical concepts, to communicate the way that an algorithm works and to convince a reader that the results will be correct..."
script = script.lower()

alphaDict = {}
for item in script:
    if item in alphaDict:
        alphaDict[item] += 1
    else:
        alphaDict[item] = 1
```

```python
sortedAlphaDict = dict(sorted(alphaDict.items(), key=lambda item: item[1], reverse=True))

print("Character frequencies (sorted by frequency):")
for char, freq in sortedAlphaDict.items():
    print(f"{char}: {freq}")
```

Part a

```python
def database(serverInfo):
    db = {}
    for item in serverInfo:
        subDic = {}
        for pair in item.split(";"):
            key, value = pair.split("=")

            if key == "id":
                value = int(value)
                
            subDic[key] = value
        db[subDic["id"]] = {
            "username": subDic["username"],
            "surname": subDic["surname"],
            "role": subDic["role"]
        }
    return db
```

```python
serverInfo = ("id=0;role=admin;username=joe;surname=naysmith", 
              "surname=suffi;username=sam;role=guest;id=421",
              "id=33;surname=lee;username=mia;role=staff")
```

```python
database(serverInfo)
```

Part b

```python
db = database(serverInfo)

for userId in db:
    user = db[userId]
    
    surname = user['surname'].capitalize()
    username = user['username'].capitalize()
    role = user['role']
    
    print(f"{surname}, {username} - {role}")
```
