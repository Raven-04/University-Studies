```python
def simplify_IP(ip):
    splitIpList = ip.split('.')
    simplifiedIps = []

    for ipPart in splitIpList:
        simplifiedIps.append(str(int(ipPart)))
    
    return '.'.join(simplifiedIps)
```

```python
print(simplify_IP("255.036.01.02"))
print(simplify_IP("128.10.0.001"))
```

```python
def formatNumber(number, format):
    numberStr = str(number)
    formatParts = format.split('+')
    formattedNumber = ""
    index = 0

    for part in formatParts:
        formattedNumber += numberStr[index:index+int(part)] + "-"
        index += int(part)

    return formattedNumber[:-1]
```

```python
phoneNumber = 971506455734
phoneFormat = "3+2+3+4"

print(f"UAE:")
print(f"phoneNumber = {phoneNumber}, phoneFormat = \"{phoneFormat}\"")
print(f"formatted output: {formatNumber(phoneNumber, phoneFormat)}")
```

```python
def firstRepeat(s):
    words = s.split(" ")
    
    if words[0] in words[1:]:
        return words[0]
    else:
        return firstRepeat(" ".join(words[1:]))
```

```python
print(firstRepeat("1 2 3 2 4 1 2 3 5"))	
print(firstRepeat("zero one and two and one"))
```
