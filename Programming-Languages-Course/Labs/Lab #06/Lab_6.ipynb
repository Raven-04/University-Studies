```python
values = [55, 12, 27, 34, 61, 49, 8]
size = 7
temp = 0
```

```python
for i in range(1, size):
    for j in range(0, size - i):
        if values[j] > values[j+1]:
            values[j], values[j+1] = values[j+1], values[j]
```

```python
print("Array after bubble sort:")

for i in range(size):
    print(" ", values[i], end="")
```

```python
def sum_list(numbers):
    sum = 0
    for num in numbers:
        if type(num) is int:
            sum += num
    return sum
```

```python
int_ls = ['1', 'two', 3.14, 4, [5], 6+3j, 7, 25//3]

sum_list(int_ls)
```

```python
import string
import random
```

```python
def generatePass(length):
    password = ''
    
    for i in range(length):
        password += random.choice(string.ascii_letters + string.digits + string.punctuation)
    
    return password
```

```python
print("Enter your password length:")
print(generatePass(10))
```

Part A

```python
def armstrong(num):
    order = len(str(num))
    sum = 0
    temp = num

    while temp > 0:
        digit = temp % 10
        sum += digit ** order
        temp //= 10

    return num == sum
```

```python
armstrong(407)
```

```python
def fib(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fib(n-1) + fib(n-2)
```

```python
fib(4)
```
